## Introduction to multistage decimation

The basic principle of oversampling is to sample a signal at an initial increased sampling rate $M\cdot f_s$, and immediately decimate by a factor $M$ to the sample rate $f_s$. Often, the initial sampling rate may be very high, and implementing a high order decimation filter to process a signal at this high smapling rate can be computationally costly. A solution to this challenge is to perform the downsampling two stages:

* Stage 1: coarse filtering with a simple lowpass filter, followed by downsampling by a factor of $M_1$
* Stage 2: higher order filtering and downsampling by a factor $M_2$, resulting in a full downsampling factor $M = M_1 \cdot M_2$

Below you can see a block diagram illustrating how such a multistage decimation system may be structured.

![](figures/multistage.png)

The advantage this offers is to introduce a wide "sacrifical band" in the intermediary stage where we don't care if aliasing occurs. As a result, we can design filter 1 to have a very wide transition band. And since required filter size is inversely proportional to transition band width, this will reduce the required filter size greatly. Once decimation stage $1$ is complete, we can use a longer filter operating at $f_{s, \small{\text{int}}}= \frac{f_{s, \small{\text{old}}}}{M_1}$, and suppress any content within the "sacrificial band". This may be a rather cumbersome concept to grasp right away, so below is shown a figure showing the frequency responses for each of the decimation filters for a two-stage system converting from  $f_{s, \small{\text{old}}} = 1200\text{ Hz}$ to $f_{s, \small{\text{new}}} = 200\text{ Hz}$.

![](figures/multistage_filtering.png)

The upper figure shows the magnitude response of the stage 1 decimation filter as a continuous blue line. Since the intended sampling frequency after decimation is $f_{s, \text{new}}=200\text{Hz}$, the sensible passband edge frequency $f_{\text{pass}}= \frac{f_{s, \text{new}}}{2} = 100\text{Hz}$. As for transition width, the fist downsampling stage reduces sampling rate to $f_{s, \text{int}} = \frac{f_{s, \text{old}}}{3} = 400\text{Hz}$, meaning spectral copies of the filter's magnitude response occur centered at at $f = \pm 400\text{Hz}$. To avoid aliasing from overlapping spectral copies in the frequency range $f \in (-f_{\text{pass}}, f_{\text{pass}})$, the filter's stopband needs to be $f_{\text{stop}} = 300 \text{Hz}$.

The lower figure shows the magnitude response of the stage 2 decimation filter as a continuous blue line. This filter is used to preserve as many frequency components as possible in the range $f \in [0\text{Hz}, 100\text{Hz})$, while suppressing as much as possible in the range $f \in [100\text{Hz}, 200\text{Hz})$ before the nyquist frequency "shrinks" to $100 \text{Hz}$. Therefore, we select a filter with a very narrow transition band, with the transition band and $-6\text{dB}$ cutoff frequency at $f_c = 100\text{Hz}$. 

So, to summarize, conducting the decimation operation des ribed above might involve a $15$-tap filter for stage 1 wich filters the signal at $f_{s, \text{old}} = 1200\text{Hz}$, and a $115$-tap filter for stage 2 which filters the signal at $f_{s, \text{int}} = 400\text{Hz}$. In contrast, if we were to do the decimation in one stage and design a decimation filter with an equally narrow passband, we would need a approximately $341$-tap filter to process the signal at $f_{s, \text{old}} = 1200\text{Hz}$ in order to do the same job.
