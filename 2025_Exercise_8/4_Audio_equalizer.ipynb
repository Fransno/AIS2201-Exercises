### Introduction to band-splitting

In this problem we will use FIR filters to construct a simple Audio Equalizer. The goal is to split an audio file sampled at $f_s = 44\ 100 \text{Hz}$ into the following frequency bands:

* **Bass:** $0\text{Hz} \leq f < 250 \text{Hz}$
* **Midrange:** $250\text{Hz} \leq f < 2000 \text{Hz}$
* **Treble:** $2\ 000\text{Hz} \leq f < 22\ 050 \text{Hz}$

This is typically accomplished by means of a filter bank. However, with FIR filters we can do some optimization, and use a single FIR lowpass filter to split a signal into a lowpass and highpass band, where the highpass band is filtered with the lowpass filter's complement. The structure of such a band-splitter is shown in the figure below:

![](figures/band_splitter_a.png)

Seeing as a $M$-tap FIR lowpass filter with a symmetrical impulse response $h_{\tiny{\text{LP}}}[n]$ imparts a delay of $\frac{M - 1}{2}$ samples on each frequency component in the passband, we can simply subtract the lowpass filtered signal from a correspondingly aligned unfiltered signal $x\left[n - \frac{M-1}{2}\right]$. One important note is that this will only work if the number of taps $M$ in the FIR filter is odd.
